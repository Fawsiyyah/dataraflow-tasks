```python
#1.write a NumPy program to test element-wise for NaN of a given array.

import numpy as np

# Create an array with numeric values and NaN entries
arr = np.array([10.5, np.nan, 3.14, np.inf, np.nan, 0.0])

# Test element-wise for NaN
result = np.isnan(arr)

print("Original array:")
print(arr)

print("\nElement-wise NaN check:")
print(result)
```

```python
#2.Write a NumPy program to test element-wise for complex number, real number of a given array. Also test whether a given number is a scalar type or not.
import numpy as np

# Create an array with mixed real and complex numbers
arr = np.array([10, 2.5 + 3j, 5, 4j, 7.8, -3])

# 1. Test element-wise for complex numbers
is_complex = np.iscomplex(arr)

# 2. Test element-wise for real numbers
is_real = np.isreal(arr)

print("Original array:")
print(arr)

print("\nElement-wise check for complex numbers:")
print(is_complex)

print("\nElement-wise check for real numbers:")
print(is_real)

# 3. Test whether a given value is a scalar type or not
scalar_val = 42
array_val = np.array([42])

print(f"\nIs {scalar_val} a scalar type? {np.isscalar(scalar_val)}")
print(f"Is the array element/array a scalar type? {np.isscalar(array_val)}")
```

```python
#3Write a NumPy program to test whether two arrays are element-wise equal within a tolerance.
arr1 = [3.0, 5, 8]
arr2 = [3, 6, 8]
# Test element-wise equality within tolerance
result = np.isclose(arr1, arr2)

print("Array 1:")
print(arr1)

print("\nArray 2:")
print(arr2)

print("\nElement-wise equality within tolerance:")
print(result)
```

```python
#4. Write a NumPy program to create an element-wise comparison (greater, greater_equal, less and less_equal) of two given arrays.
## Define two given arrays
array1 = np.array([1, 5, 3, 9])
array2 = np.array([3, 5, 2, 7])

# Perform element-wise comparisons
greater_result = np.greater(array1, array2)
greater_equal_result = np.greater_equal(array1, array2)
less_result = np.less(array1, array2)
less_equal_result = np.less_equal(array1, array2)

print("Array 1:")
print(array1)

print("\nArray 2:")
print(array2)

print("\nElement-wise greater:")
print(greater_result)

print("\nElement-wise greater or equal:")
print(greater_equal_result)

print("\nElement-wise less:")
print(less_result)

print("\nElement-wise less or equal:")
print(less_equal_result)
```

```python
#Write a NumPy program to create an element-wise comparison (equal, equal within a tolerance) of two given arrays.

# Define two given arrays
array1 = np.array([1.0, 2.0, 3.0, 4.0])
array2 = np.array([1.0, 2.0001, 3.0, 4.005])

# Element-wise exact comparison
equal_result = np.equal(array1, array2)

# Element-wise comparison within a tolerance
is_close_result = np.isclose(array1, array2, rtol=1e-02, atol=1e-02)

print("Array 1:")
print(array1)

print("\nArray 2:")
print(array2)

print("\nElement-wise equal:")
print(equal_result)

print("\nElement-wise equal within a tolerance:")
print(is_close_result)
```
