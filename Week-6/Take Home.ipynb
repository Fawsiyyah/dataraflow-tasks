```python
#taking a users birth year as input and calculating thir age
from datetime import date

# Get birthdate from the user
birth_year = int(input("Enter your birth year (YYYY): "))
birth_month = int(input("Enter your birth month (MM): "))
birth_day = int(input("Enter your birth day (DD): "))

# Creating date objects for birthdate and today
birth_date = date(birth_year, birth_month, birth_day)
today = date.today()

# Calculate age
age = today.year - birth_date.year

# Adjust if birthday has not occurred yet this year
if (today.month, today.day) < (birth_date.month, birth_date.day):
    age -= 1

print(f"You are {age} years old.")
```

```python
import json

#Step 1: Defining a list of student records
students = [
    {"name": "Peter", "age": 20, "grade": "A"},
    {"name": "Paul", "age": 22, "grade": "B"},
    {"name": "Charles", "age": 21, "grade": "D"}
]

file_name = "students.json"

# Write the list of records to a JSON file
with open(file_name, "w") as file:
    json.dump(students, file, indent=4)

print(f"Data successfully written to {file_name}")

# Read the JSON file back into a Python object
with open(file_name, "r") as file:
    loaded_students = json.load(file)

print("\nData read back from JSON file:")
print(loaded_students)
print(f"Type of loaded data: {type(loaded_students)}")
```

```python
import random

def math_quiz():
    print("=== Welcome to the Math Quiz! ===")
    print("You will face 5 random questions. Good luck!\n")
    
    score = 0
    total_questions = 5
    operators = ['+', '-', '*', '/']

    for i in range(1, total_questions + 1):
        # Generate random numbers and a random operator
        num1 = random.randint(1, 12)
        num2 = random.randint(1, 12)
        op = random.choice(operators)
        
        # Ensuring division results in a clean integer
        if op == '/':
            num1 = num1 * num2  # Guarantees num1 / num2 is an integer

        # Calculate the correct answer
        if op == '+':
            correct_answer = num1 + num2
        elif op == '-':
            correct_answer = num1 - num2
        elif op == '*':
            correct_answer = num1 * num2
        elif op == '/':
            correct_answer = num1 // num2  # Use integer division since it's clean

        # Ask the user the question
        print(f"Question {i}: What is {num1} {op} {num2}?")
        
        # Get user input and handle invalid inputs gracefully
        try:
            user_answer = int(input("Your answer: "))
        except ValueError:
            print("Invalid input. Counting as incorrect.")
            user_answer = None

        # Check the answer
        if user_answer == correct_answer:
            print("Correct!\n")
            score += 1
        else:
            print(f"Incorrect. The correct answer was {correct_answer}.\n")

    # Display final score
    print("=== Quiz Complete ===")
    print(f"Your final score: {score} out of {total_questions}")

# Run the program
if __name__ == "__main__":
    math_quiz()
```

```python
def divide_one_hundred():
    print("=== Division Calculator ===")
    user_input = input("Enter a number to divide 100 by: ")
    
    try:
        # Convert input to a float to allow both integers and decimals
        number = float(user_input)
        
        # Perform the division
        result = 100 / number
        
    except ZeroDivisionError:
        print("Error: You cannot divide by zero!")
        
    except ValueError:
        print("Error: Invalid input. Please enter a valid numerical value.")
        
    else:
        # This block runs only if no exceptions were raised
        # Formatting to 2 decimal places if it's a float
        print(f"Success: 100 divided by {number} is {result:.2f}")
        
    finally:
        print("==========================")

# Run the program
if __name__ == "__main__":
    divide_one_hundred()
```

```python
from datetime import datetime, date

def days_until_birthday():
#birtday countdown:
    
    # Get user input for the birthday
    birthday_input = input("Enter your birthday (YYYY-MM-DD): ").strip()
    
    try:
        # Parse the input string into a date object
        birth_date = datetime.strptime(birthday_input, "%Y-%m-%d").date()
        today = date.today()
        
        # Create a candidate for the next birthday in the current year
        next_birthday = date(today.year, birth_date.month, birth_date.day)
        
        # If the birthday has already passed this year, look to next year
        if next_birthday < today:
            next_birthday = date(today.year + 1, birth_date.month, birth_date.day)
            
        # Calculate the delta
        days_left = (next_birthday - today).days
        
        # Print results
        if days_left == 0:
            print(f"🎉 **Happy Birthday!** Today is the big day!")
        else:
            print(f"🗓️  There are **{days_left} days** left until your next birthday ({next_birthday.strftime('%B %d, %Y')}).")
            
    except ValueError:
        print("Error: Invalid date format. Please use **YYYY-MM-DD** (e.g., 1995-05-18).")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")

if __name__ == "__main__":
    days_until_birthday()
```

```python
def convert_string_to_int():
    invalid_string = "abc"
    
    try:
        # This will fail because "abc" cannot be converted into base-10 digits
        number = int(invalid_string)
        print(f"Success! The number is {number}")
        
    except ValueError as error:
        # Catches the specific value error and prints a custom message
        print("Error: Conversion failed!")
        print(f"Details: {error}")

if __name__ == "__main__":
    convert_string_to_int()
```

```python
#Write a python program that asks the user for two numbers and prints their greatest common divisor (GCD).
import math

def main():
    
    try:
        # Ask the user for two integers
        num1 = int(input("Enter the first integer: "))
        num2 = int(input("Enter the second integer: "))
        
        # Calculate the GCD using Python's built-in math module
        result = math.gcd(num1, num2)
        
        # Print the result
        print(f"The GCD of {num1} and {num2} is: {result}")
        
    except ValueError:
        print("Invalid input! Please enter whole numbers (integers) only.")

if __name__ == "__main__":
    main()
```

```python
from datetime import datetime

def get_date_input(prompt):
    while True:
        date_str = input(prompt)
        try:
            # Parse the string into a datetime object
            return datetime.strptime(date_str, "%Y-%m-%d").date()
        except ValueError:
            print("Invalid format! Please use YYYY-MM-DD (e.g., 2026-09-26).")

def main():
    print("Please enter dates in YYYY-MM-DD format.\n")
    
    # Get the two dates from the user
    date1 = get_date_input("Enter the first date: ")
    date2 = get_date_input("Enter the second date: ")
    
    # Calculate the absolute difference in days
    delta = abs(date2 - date1)
    
    print(f"\n📅 The total number of days between {date1} and {date2} is: {delta.days} days.")

if __name__ == "__main__":
    main()
```

```python
def get_integer_input():
    
    while True:
        user_input = input("Please enter a whole number (integer): ")
        
        try:
            # Attempt to convert the input string to an integer
            valid_integer = int(user_input)
            
            # If the line above doesn't crash, the input is valid!
            print(f"Success! You entered the integer: {valid_integer}")
            return valid_integer
            
        except ValueError:
            # This block runs if int() fails (e.g., if the user typed text or a decimal)
            print(f" '{user_input}' is not a valid integer. Please try again.\n")

if __name__ == "__main__":
    get_integer_input()
```
