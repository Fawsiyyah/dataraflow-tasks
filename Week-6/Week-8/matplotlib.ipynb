```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6, 7] #weekdays
y = [964, 287, 42, 390, 64, 800, 523]  #viewsperday

plt.plot(x, y, label="Youtube views", 
         color='r', 
         marker='o', 
         markerfacecolor = 'g', 
         linestyle='--',
         linewidth=3)

plt.xlabel("Week days")
plt.ylabel("Views")
plt.title("Youtube views on daily basis")
plt.legend(loc = "upper left")
plt.show()
```

```python
#bar charts

import matplotlib.pyplot as plt

y_views=[200, 50, 250, 300, 100, 150,350]
fb_views=[160, 200, 40, 240, 80, 120,280]
days=[1,2,3,4,5,6,7]

plt.bar(days, y_views, label="Youtube views")
plt.bar(days, fb_views, label="Facebook views")

plt.xlabel('Days')
plt.ylabel('Views')
plt.title('Social media views for a Channel')
plt.legend()
plt.savefig('bar_2.png', dpi = 500)
plt.show()
```

```python
# Histograms

import matplotlib.pyplot as plt

ages=[22,55,62,45,21,22,99,34,42,4,102,110,27,48,99,84]
bins=[0,20,40,60,80,100,120]

plt.hist(ages, bins, histtype='bar', rwidth=0.8)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.title('Current ages of people')

plt.savefig('hist.png', dpi = 500)
plt.show()
```

```python
#pie chart

import matplotlib.pyplot as plt

labels_1 = ['Facebook', 'Instagram', 'Youtube', 'LinkedIn']
views = [300, 350, 400, 450]

explode_1 = (0, 0, 0, 0.2) #explode first slice

plt.pie(views, labels=labels_1, autopct = '%1.1f%%', explode = explode_1, shadow = True)
plt.title('Social Media Views')
plt.savefig('pie.png', dpi = 500)
plt.show()
```
