````python
class BankAccount:
    def __init__(self, owner, balance = 0.0 ):
        self.owner = owner
        self.balance = balance
    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount:.2f}. New Balance: ${self.balance:.2f}")
        else:
            print("Deposit amount must be positive.")
    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew ${amount:.2f}. New Balance: ${self.balance:.2f}")
        elif amount > self.balance:
            print(f"Failed withdrawal of ${amount:.2f}: Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

        # Subclass SavingsAccount inheriting from BankAccount
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0.0, interest_rate=0.02):
        # Initialize parent attributes using super()
        super().__init__(owner, balance)
        # Add new attribute unique to SavingsAccount (e.g., 0.02 = 2%)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Interest of ${interest:.2f} applied ({self.interest_rate * 100}%). New Balance: ${self.balance:.2f}")


# 1. Instantiate a SavingsAccount
account = SavingsAccount(owner="Zeee", balance=1000.0, interest_rate=0.05)
print(f"Account created for {account.owner} with initial balance: ${account.balance:.2f}")

# 2. Demonstrate standard BankAccount methods
account.deposit(500.0)      
account.withdraw(200.0)    

# 3. Demonstrate unique SavingsAccount method
account.add_interest()      

# 4. Demonstrate a failed withdrawal check
account.withdraw(2000.0)    
        
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
#Creating the CountDown class 
class CountDown:
    def __init__(self, n):
        self.current = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        
        number = self.current
        self.current -= 1
        return number

#Demonstration in a for loop

print("Starting Countdown:")
# Instantiating the iterator with a start value of 5
for num in CountDown(5):
    print(num)
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
import os

# Creating custom Iterator Class
class FileLineReader:
    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def __iter__(self):
        # Open the file when iteration begins
        self.file = open(self.filename, 'r', encoding='utf-8')
        return self

    def __next__(self):
        # Read the next line
        line = self.file.readline()
        
        # readline() returns an empty string "" only when EOF (End of File) is reached
        if line == "":
            self.file.close()  
            raise StopIteration
            
        # Return the line, stripping trailing newlines for clean printing
        return line.strip()


# --- Setup Sample File and Demonstration ---

sample_filename = "sample_data.txt"

#Creating a dummy text file to test with
with open(sample_filename, "w", encoding="utf-8") as f:
    f.write("Line 1: Python OOP is powerful.\n")
    f.write("Line 2: Iterators save memory.\n")
    f.write("Line 3: Reading file one line at a time.")

#Using the custom iterator in a for loop
print("Reading file content using FileLineReader:")
try:
    for line in FileLineReader(sample_filename):
        print(line)
finally:
    # Clean up the sample file from your system
    if os.path.exists(sample_filename):
        os.remove(sample_filename)
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci():
    def __init__(self, n = int):
        self.max_value = n
        self.a = 0
        self.b = 1
    def __iter__ (self):
        return self
    def __next__(self):
        if self.a > self.max_value:
            raise StopIteration

        #save the current value, then return the sequence
        current = self.a
        self.a, self.b = self.b, self.a + self.b
        return current
for num in Fibonacci(15):
 print(num, end=" ")
````

````python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        #Track availability status
        self.is_borrowed = False

    def get_description(self):
        return f"'{self.title}' by {self.author} ({self.year})"
    
class EBook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        base_desc = super().get_description()
        return f"{base_desc} [E-Book: {self.format}, {self.file_size}]"

class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimension):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimension = dimension


    def get_description(self):
        base_desc = super().get_description()
        return f"{base_desc} [Printed: {self.weight}, {self.dimensions}]"

#Member
class Member:
    # Class-level variable tracking total books borrowed across all members
    total_borrowed_books = 0

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        # Protected attribute (indicated by a single underscore)
        self._borrowed_books = []
    # Getter for protected attribute
    @property
    def borrow_book(self, book):
           # Return a copy to prevent unintended external modification
            return list(self._borrowed_books)
    def borrow_book(self, book):
        """Base method to be overridden by subclasses with specific limits."""
        raise NotImplementedError("Subclasses must implement borrow_book limits")

    def return_book(self, book):
        if book in self._borrowed_books:
            self._borrowed_books.remove(book)
            book.is_borrowed = False
            Member.total_borrowed_books += 1
            print(f"{self.name} returned: {book.title}")
            return True
        print(f"{self.name} does not have '{book.title}' borrowed.")
        return False
    
#member iteration
    def __iter__(self):
        return iter(self.borrowed_books)

class StudentMember(Member):
    book_limit = 5
    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level
        
    def borrow_book(self, book):
        if len(self._borrowed_books) >= self.book_limit:
            print(f"Limit reached! Student {self.name} can borrow at most {self.book_limit} books.")
            return False
        self._borrowed_books.append(book)
        book.is_borrowed = True
        Member.total_borrowed_books += 1
        print(f"Student {self.name} successfully borrowed: {book.title}")
        return True
        
    
class TeacherMember(Member):
    book_limit = 10

    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department
        book.is_borrowed = True

    def borrow_book(self, book):
        if book.is_borrowed:
            print(f"Unavailable: '{book.title}' is already borrowed.")
            return False
        
        if len(self._borrowed_books) >= self.book_limit:
            print(f"Limit reached! Teacher {self.name} can borrow at most {self.book_limit} books.")
            return False
        self._borrowed_books.append(book)
        book.is_borrowed = True
        Member.total_borrowed_books += 1
        print(f"Teacher {self.name} successfully borrowed: {book.title}")
        return True

#library iteration
class Library:
    def __init__(self, library_name):
        self.library_name = library_name
        self.books = []

    # Getter method
    @property
    def library_name(self):
        return self.__library_name
    
    # Setter method with validation
    @library_name.setter
    def library_name(self, new_name):
        if not isinstance(new_name, str) or not new_name.strip():
            raise ValueError("Library name must be a non-empty string.")
        self.__library_name = new_name.strip()

    def add_book(self, book):
        self.books.append(book)

    def __iter__(self):
        return iter(self.books)
    
# Filter available vs borrowed books
    def get_available_books(self):
        return [book for book in self.books if not book.is_borrowed]
    def get_borrowed_books(self):
        return [book for book in self.books if book.is_borrowed]

#Case-insensitive search by title or author
    def search_books(self, query):
        q = query.lower()
        return [
            book for book in self.books
            if q in book.title.lower() or q in book.author.lower()
        ]

if __name__ == "__main__":
    lib = Library("Metropolitan Public Library")

    b1 = EBook("Fluent Python", "Luciano Ramalho", 2021, "20MB", "EPUB")
    b2 = PrintedBook("Clean Code", "Robert C. Martin", 2008, "650g", "7x9 in")
    b3 = Book("The Pragmatic Programmer", "Andy Hunt", 1999)

    lib.add_book(b1)
    lib.add_book(b2)
    lib.add_book(b3)

    student = StudentMember("Zeee", "S101", "10th Grade")
    teacher = TeacherMember("Mr. James", "T101", "Computer Science")

    #Borrowing books and updating availability
    student.borrow_book(b1)
    teacher.borrow_book(b2)

    # Attempting to borrow an already borrowed book
    teacher.borrow_book(b1)

    #View Available vs Borrowed books
    print("\n--- Available Books ---")
    for b in lib.get_available_books():
        print(f"- {b.title}")

    print("\n--- Borrowed Books ---")
    for b in lib.get_borrowed_books():
        print(f"- {b.title}")

    #Search books by title or author
    print("\n--- Search Results for 'python' ---")
    for result in lib.search_books("python"):
        print(f"- {result.title} by {result.author}")

    print("\n--- Search Results for 'Martin' ---")
    for result in lib.search_books("Martin"):
        print(f"- {result.title} by {result.author}")

    #Total borrowed count across all members (class-level variable)
    print(f"\nTotal books borrowed globally: {Member.total_borrowed_books}")

    # Returning a book updates the count
    student.return_book(b1)
    print(f"Total books borrowed after return: {Member.total_borrowed_books}")

    # Demonstration of Scope and Access
if __name__ == "__main__":
    lib = Library("City Central Library")
    book = Book("1984", "George Orwell", 1949)
    student = StudentMember("Zeee", "S101", "10th Grade")

    # Accessing Private Attribute via Getter / Setter
    print("Library Name (via getter):", lib.library_name)
    lib.library_name = "Metropolitan Public Library"
    print("Updated Library Name:", lib.library_name)

    # Direct access triggers AttributeError due to name mangling:
    try:
        print(lib.__library_name)
    except AttributeError as e:
        print("Direct access to __library_name failed:", e)

    #Accessing Protected Attribute via Subclass and Getter
    student.borrow_book(book)
    print("Borrowed Books count (via getter):", len(student._borrowed_books))

````
